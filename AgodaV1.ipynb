```python
import time
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager

# --- 設定區 ---
HOTEL_URL = "https://www.agoda.com/zh-hk/grand-hotel/hotel/taipei-tw.html"
TARGET_COUNT = 5  # 你想要抓取的有效評論數量

def start_crawler(url):
    options = webdriver.ChromeOptions()
    # 基礎防偵測設定
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_experimental_option("excludeSwitches", ["enable-automation"])
    options.add_experimental_option('useAutomationExtension', False)
    
    driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)
    driver.execute_script("Object.defineProperty(navigator, 'webdriver', {get: () => undefined})")
    
    try:
        print(f"🌐 正在前往飯店頁面...")
        driver.get(url)
        wait = WebDriverWait(driver, 20)

        # 1. 模擬真人滾動，確保評論區塊被觸發載入
        print("📜 正在滾動頁面載入評論區塊 (Lazy Loading)...")
        for i in range(20): # 增加滾動深度確保抓到足夠數量
            driver.execute_script(f"window.scrollTo(0, {1500 + i*600});")
            time.sleep(1)

        # 2. 定位所有評論容器
        card_selector = '[data-element-name="review-comment"]'
        wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, card_selector)))
        all_cards = driver.find_elements(By.CSS_SELECTOR, card_selector)
        
        valid_reviews = []
        print(f"🔍 偵測到 {len(all_cards)} 個潛在區塊，開始篩選旅客評論...")

        # 3. 遍歷並篩選有效評論
        valid_reviews = []
        print(f"🔍 偵測到 {len(all_cards)} 個潛在區塊，正在深度篩選...")

        for card in all_cards:
            # 如果已經抓滿 5 則就停止
            if len(valid_reviews) >= 5:
                break
                
            try:
                # --- 關鍵修正 1：排除飯店回覆區塊 ---
                # 飯店回覆的 class 通常包含 "Review-response"，我們直接排除
                if "Review-response" in card.get_attribute("class"):
                    continue

                # --- 關鍵修正 2：確保是旅客評論 (必須有分數) ---
                try:
                    score_el = card.find_element(By.CSS_SELECTOR, ".Review-comment-leftScore")
                    score = score_el.text
                except:
                    continue # 沒分數的一律視為非旅客評論，跳過

                # A. 評論 ID
                r_id = card.get_attribute("data-review-id") or "N/A"
                
                # --- 關鍵修正 3：防止重複抓取同一則 ID ---
                if any(r['id'] == r_id for r in valid_reviews):
                    continue

                # B. 用戶名稱
                try:
                    name = card.find_element(By.CSS_SELECTOR, '[data-info-type="reviewer-name"]').text
                except:
                    name = "隱名旅客"

                # C. 入住時間
                try:
                    stay_date = card.find_element(By.CSS_SELECTOR, '[data-info-type="stay-detail"]').text
                except:
                    stay_date = "未標示"

                # D. 發表日期 (從 bubble 區域找包含日期的 span)
                comment_date = "未知日期"
                try:
                    spans = card.find_elements(By.TAG_NAME, "span")
                    for s in spans:
                        if "202" in s.text and ("月" in s.text or "評價" in s.text):
                            comment_date = s.text
                            break
                except:
                    pass

                # E. 評論內容
                try:
                    # 優先抓取內容文字，避開飯店回覆的文字
                    content = card.find_element(By.CSS_SELECTOR, ".Review-comment-bodyText").text.replace('\n', ' ').strip()
                except:
                    content = "(無文字內容)"

                if content: # 確保有內容才加入
                    valid_reviews.append({
                        "id": r_id, "name": name, "score": score,
                        "stay": stay_date, "date": comment_date, "content": content
                    })

            except Exception:
                continue

        # 4. 格式化輸出結果
        print("\n" + "★" * 25 + " 抓取結果 " + "★" * 25)
        if not valid_reviews:
            print("❌ 未能成功篩選出有效評論，請檢查頁面是否正確載入。")
        else:
            for i, res in enumerate(valid_reviews):
                print(f"【評論 {i+1}】")
                print(f"🆔 評論 ID：{res['id']}")
                print(f"👤 用戶名稱：{res['name']}  | ⭐ 評分分數：{res['score']}")
                print(f"📅 發表日期：{res['date']}")
                print(f"🗓️ 入住時間：{res['stay']}")
                print(f"💬 評論內容：{res['content'][:200]}...") # 顯示前 200 字
                print("-" * 60)
        print("★" * 58 + "\n")

    except Exception as e:
        print(f"❌ 程式執行發生錯誤: {e}")
    finally:
        if 'driver' in locals():
            input("🎯 抓取完成！請確認終端機內容後，按 Enter 鍵關閉瀏覽器...")
            driver.quit()

if __name__ == "__main__":
    start_crawler(HOTEL_URL)
```

```python
import time
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager

test_url = "https://www.agoda.com/zh-hk/grand-hotel/hotel/taipei-tw.html"

try:
    print("🚀 啟動 Chrome...")
    driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
    driver.get(test_url)
    wait = WebDriverWait(driver, 20)

    print("📜 正在深度滾動以載入更多評論...")
    # 增加滾動次數（從 8 次增加到 15 次），確保頁面跑得夠深
    for i in range(15):
        # 每次滾動 800 像素
        driver.execute_script(f"window.scrollTo(0, {1500 + i*800});")
        time.sleep(1.2) # 給予一點時間讓資料從伺服器傳回來

    print("🔍 嘗試提取前 10 則評論文字...")
    
    review_selector = ".Review-comment-bodyText"
    
    # 等待評論標籤出現
    wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, review_selector)))
    
    # 抓取所有已經載入出來的評論文字標籤
    reviews = driver.find_elements(By.CSS_SELECTOR, review_selector)

    if len(reviews) > 0:
        print("-" * 50)
        # 這裡將限制數量改為 10
        actual_count = min(len(reviews), 10)
        print(f"✅ 成功找到 {len(reviews)} 則標籤，輸出前 {actual_count} 則：\n")
        
        for i in range(actual_count):
            content = reviews[i].text.strip().replace('\n', ' ')
            if content:
                print(f"【評論 {i+1}】: {content}\n")
        print("-" * 50)
    else:
        print("❌ 找不到評論，可能滾動深度不足或網路較慢。")

except Exception as e:
    print(f"❌ 發生錯誤: {e}")

finally:
    if 'driver' in locals():
        input("請確認是否已抓到 10 則，按 Enter 鍵關閉...")
        driver.quit()
```

```python
import time
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager

# 設定目標：你要抓取的飯店網址與數量
HOTEL_URL = "https://www.agoda.com/zh-hk/grand-hotel/hotel/taipei-tw.html"
MAX_REVIEWS = 10 

def run_agoda_spider(url, target_count):
    options = webdriver.ChromeOptions()
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_experimental_option("excludeSwitches", ["enable-automation"])
    
    driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)
    driver.execute_script("Object.defineProperty(navigator, 'webdriver', {get: () => undefined})")
    
    all_results = []
    seen_ids = set()

    try:
        driver.get(url)
        wait = WebDriverWait(driver, 20)

        while len(all_results) < target_count:
            print(f"📜 正在處理頁面，目前已成功抓取 {len(all_results)} 則...")
            
            # 1. 深度滾動，確保評論區塊與分頁按鈕載入
            for _ in range(10):
                driver.execute_script("window.scrollBy(0, 800);")
                time.sleep(0.8)
            
            # 2. 定位所有評論卡片 (根據截圖中的 data-element-name)
            card_selector = '[data-element-name="review-comment"]'
            wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, card_selector)))
            cards = driver.find_elements(By.CSS_SELECTOR, card_selector)

            # 3. 解析當前頁面的卡片
            for card in cards:
                if len(all_results) >= target_count: break
                
                try:
                    # 排除飯店回覆：檢查是否有分數標籤
                    score_el = card.find_elements(By.CSS_SELECTOR, ".Review-comment-leftScore")
                    if not score_el: continue
                    
                    # 檢查 ID 防止重複抓取
                    r_id = card.get_attribute("data-review-id")
                    if not r_id or r_id in seen_ids: continue
                    seen_ids.add(r_id)

                    # 提取各欄位資料 (依據你的截圖定位)
                    score = score_el[0].text
                    name = card.find_element(By.CSS_SELECTOR, '[data-info-type="reviewer-name"]').text
                    stay_date = card.find_element(By.CSS_SELECTOR, '[data-info-type="stay-detail"]').text
                    content = card.find_element(By.CSS_SELECTOR, ".Review-comment-bodyText").text.replace('\n', ' ').strip()
                    
                    # 發表日期：搜尋包含「評價」或「202」字眼的 span
                    comment_date = "未知日期"
                    spans = card.find_elements(By.TAG_NAME, "span")
                    for s in spans:
                        if "202" in s.text and ("月" in s.text or "評價" in s.text):
                            comment_date = s.text
                            break

                    all_results.append({
                        "id": r_id, "name": name, "score": score,
                        "stay": stay_date, "date": comment_date, "content": content
                    })
                except:
                    continue

            if len(all_results) >= target_count: break

            # 4. 執行翻頁 (使用你截圖中的下一個評論頁面按鈕)
            print("➡️ 嘗試點擊下一頁按鈕...")
            try:
                next_btn_selector = 'button[data-element-name="review-paginator-next"]'
                next_btn = driver.find_element(By.CSS_SELECTOR, next_btn_selector)
                
                # 滾動到按鈕位置並強制點擊
                driver.execute_script("arguments[0].scrollIntoView({block: 'center'});", next_btn)
                time.sleep(1)
                driver.execute_script("arguments[0].click();", next_btn)
                
                time.sleep(5) # 等待 AJAX 載入新內容
            except Exception as e:
                print(f"🛑 無法翻頁或已達末頁")
                break

        # 5. 最終終端機輸出
        print("\n" + "★" * 20 + f" 抓取完成：共 {len(all_results)} 則 " + "★" * 20)
        for i, res in enumerate(all_results):
            print(f"【第 {i+1} 則】 ID: {res['id']}")
            print(f"👤 用戶: {res['name']}  |  ⭐ 分數: {res['score']}")
            print(f"📅 發表: {res['date']}  |  🗓️ 入住: {res['stay']}")
            print(f"💬 內容: {res['content'][:150]}...")
            print("-" * 60)

    except Exception as e:
        print(f"❌ 執行出錯: {e}")
    finally:
        if 'driver' in locals():
            input("\n確認完畢請按 Enter 關閉瀏覽器...")
            driver.quit()

if __name__ == "__main__":
    run_agoda_spider(HOTEL_URL, MAX_REVIEWS)
```
